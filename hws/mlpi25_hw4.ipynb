#### P1. (5 points)

Use a decision tree algorithm in scikit-learn to classify a simple XOR dataset used in lecture 4. Split the data to 70% training and 30% testing. You must accomplish the following:

1) Plot the resulting decision boundary.
2) Use tree.plot_tree() in Scikit-learn to display the decision tree diagram.
3) Evaluate train and test accuracy.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(1)
X_xor = np.random.randn(200, 2)
y_xor = np.logical_xor(X_xor[:, 0] > 0,
                       X_xor[:, 1] > 0)
y_xor = np.where(y_xor, 1, 0)

plt.scatter(X_xor[y_xor == 1, 0],
            X_xor[y_xor == 1, 1],
            c='royalblue',
            marker='s',
            label='Class 1')
plt.scatter(X_xor[y_xor == 0, 0],
            X_xor[y_xor == 0, 1],
            c='tomato',
            marker='o',
            label='Class 0')

plt.xlim([-3, 3])
plt.ylim([-3, 3])
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

plt.legend(loc='best')
plt.tight_layout()
plt.show()
```
